## Q5 — Create Inventory Value Column

Add a new column:

inventory_value = quantity * price

Then display:

product name
category
inventory_value

Concepts:

withColumn()
transformations
arithmetic operations


```python
prod_new = prod.withColumn("inventory_value", prod.quantity * prod.price)
prod_new.select('name','Category','inventory_value').show()
prod_new.show()
```

## Q6 — Create Discounted Price Column

Create a new column:

discounted_price

Rules:

Electronics → 15% discount
Clothing → 10% discount
Books → 5% discount
Others → 8% discount

Concepts:

when()
otherwise()
conditional columns
withColumn()


```python
prod_discount = prod.withColumn( "discounted_price", 
                             when(col("category") == "Electronics", col("price") * 0.85) 
                                 .when(col("category") == "Clothing", col("price") * 0.90) 
                                 .when(col("category") == "Books", col("price") * 0.95) 
                                 .otherwise(col("price") * 0.92) ) 
prod_discount.show()
```

## Q7 — Add Stock Status Column

Create a new column:

stock_status

Conditions:

quantity < 10 → "Low Stock"
quantity between 10 and 30 → "Medium Stock"
quantity > 30 → "High Stock"

Concepts:

conditional transformations
multiple conditions


```python
prod_stock = prod.withColumn("stock_status", 
                             when(col("quantity") < 10, "Low Stock") 
                                 .when((col("quantity") >= 10) & (col("quantity") <= 30), "Medium Stock") 
                                 .otherwise("High Stock" )) 
prod_stock.show()
```

## Q8 — Convert Data Types

Convert:

quantity → IntegerType
price → DoubleType

Then print the schema.

Concepts:

cast()
printSchema()


```python
prod_cast = prod.withColumn( "quantity", 
                         col("quantity")
                             .cast(IntegerType()) ).withColumn( "price", 
                                                                col("price").cast(DoubleType()) ) 
prod_cast.printSchema()
```

## Q11 — Category Analytics

For each category find:

total number of products
average price
total inventory value

Concepts:

groupBy()
count()
avg()
sum()
aggregations

```python
prod.groupby("category").agg(
    count("*").alias("Total_Products"),
    avg("price").alias("avg_price"),
    sum(col("quantity")*col("price")).alias("inventory_value")).show()
```

## Q12 — Clean Null Values

Assume:

some quantity values are null
some price values are null

Tasks:

replace null quantity with 0
replace null price with average price

Concepts:

fillna()
null handling
aggregations

```python
avg_price = prod.select(avg("price")).collect()[0][0] 
df_clean = prod.fillna({ "quantity": 0 }) 
df_clean = df_clean.fillna({ "price": avg_price }) 
df_clean.show()
```
