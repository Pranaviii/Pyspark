# Filter the rows based on column values in PySpark dataframe.

## Building Dataframe

```python
# importing sparksession from pyspark.sql module
from pyspark.sql import SparkSession
```

```python
# creating sparksession and giving an app name
spark = SparkSession.builder \
    .appName("sparkdf") \
    .getOrCreate()
```

```python
spark
```

```python
# list  of employee data
data = [["1", "sravan", "Vignan"],
        ["2", "ojaswi", "vvit"],
        ["3", "rohith", "vvit"],
        ["4", "sridevi", "Vignan"],
        ["1", "sravan", "iit"],
        ["4", "sridevi", "Vignan"],
        ["5", "gnanesh", "iit"]]

# specify column names
columns = ['ID', 'NAME', 'College']
```

```python
# creating a dataframe from the lists of data
dataframe = spark.createDataFrame(data, columns)

dataframe.show()
```

### Filter rows in dataframe where ID =1

```python
dataframe.where(dataframe.ID == '1').show()
```

### Get the data where name not 'sravan'

```python
dataframe.where(dataframe.NAME != 'sravan').show()
```

### Filter rows where ID greater than 2 and college is vvit

```python
dataframe.where((dataframe.ID > '2') & (dataframe.College == 'vvit')).show()
```

### column value = vvit college

```python
dataframe.filter(dataframe.College == 'vvit').show()
```

### filter the data where id > 3.

```python
# get the data where id > 3
dataframe.filter(dataframe.ID>'3').show()
```

### Filter rows where ID greater than 2 and college is vignan

```python
dataframe.filter((dataframe.ID>'2') & (dataframe.College == 'Vignan')).show()
```

### Find data with 2,3 ids or iit, vignan college

```python
dataframe.filter((dataframe['ID'].isin([2,3])) | (dataframe['College'].isin(['iit', 'Vignan']))).show()
```

### Drop duplicates

```python
dataframe.dropDuplicates(["Name"]).show()
```

```python
# data count
dataframe.count()
```

```python
# sort
dataframe.sort('ID').show()
```

```python
dataframe.sort("ID", ascending=False).show()
```
