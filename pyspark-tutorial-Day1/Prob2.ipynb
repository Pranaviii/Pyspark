# Read CSV File

```python
from pyspark.sql import SparkSession
```

```python
# Create a SparkSession
spark = SparkSession.builder.appName("Prob2").getOrCreate()
```

```python
spark
```

```python

prod = spark.read.load("./data/products.csv",format="csv", sep=",", inferSchema="true", header="true")
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, DoubleType
```

```python
prod.show()
prod.printSchema()
```

```python
# Import the module
from pyspark.sql.functions import *

prod.groupBy(prod["Category"]).agg(sum("price")).show()
prod.groupBy(prod["Category"]).agg(sum("price").alias('Total price')).show()
```

## Q1 — Select Specific Columns

Display only:

name
category
price

select()

```python
prod.select('name','Category','price').show()
```

## Q2 — Filter Expensive Products

Show all products where:

price > 100

Practice:

filter()

```python
prod.filter(prod.price > '100').show()
```

## Q3 — Filter Electronics Products

Display only products from:

"Electronics"

Practice:

conditional filtering

```python
prod.where(prod.category == 'Electronics').show()
```

## Q4 — Sort Products by Price

Display all products sorted:

from highest price to lowest price

Concepts:

orderBy()
sorting


```python
prod.orderBy('category').sort("price", ascending=False).show()
```
